### Aufgabe 1.1 b)

Der [Hamming-Abstand](https://de.wikipedia.org/wiki/Hamming-Abstand) von zwei Binärcodes  $x = (x_1, \dots, x_n )$ mit $x_i \in \{0,1\} $ und $y = (y_1, \dots, y_n)$ mit $y_i \in \{0, 1\}$ ist definiert als:

$ h(x, y) := | \{j \in \{1, ... , n\} | x_j  \neq y_j \} | $

<br>

Implementieren Sie die Funktion `random_binary`, die zufällige Binärcodes der Länge $n$ mithilfe des `random`-Pakets generiert und generieren Sie zwei zufällige Codes der Länge $n=10$. 

Berechnen Sie anschließend mit der Funktion `hamming_distance` den Hamming-Abstand dieser Binärcodes. 

*Hinweis:* Benutzen Sie die Funktion `zip`. Eine Beschreibung der Funktion erhalten Sie durch `print(zip.__doc__)`

```python
import random

def random_binary(length: int) -> List[int]:  # 1 Punkt
    # ToDo
    return [random.randint(0, 1) for _ in range(length)]

print("Random Binary codes:")
a = random_binary(10)
b = random_binary(10)

print(a)
print(b)
print()

def hamming_distance(bincode0: List[int], bincode1: List[int]) -> int:  # 1 Punkt
    # ToDo
    return sum([1 for x, y in zip(bincode0, bincode1) if x != y])


print(f" => Distance: {hamming_distance(a, b)}")
```

### Aufgabe 1.1 c) 

Implementieren Sie die Funktion `avg_hamming_distance`, die zufällige Binärpaare erzeugt und den durchschnittlichen Hamming-Abstand aller Paare berechnet. Berechnen Sie damit den durchschnittlichen Abstand von 10000 Binärpaaren der Länge $n=10$.

```python
def avg_hamming_distance(n: int, length: int) -> float:  # 1 Punkt
    # ToDo
    distances = [hamming_distance(random_binary(length), random_binary(length))
    for _ in range(n)]
    return sum(distances) / n

n = 10000
l = 10

print(f"Average hamming distance for {n} binary pairs of length {l}: {avg_hamming_distance(n, l)}")
```
